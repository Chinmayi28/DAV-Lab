```python
import numpy as np
import pandas as pd
```

```python
file_path=(r"C:\Users\Prakash\Downloads\cafe_sale\dirty_cafe_sales.csv")
df=pd.read_csv(file_path)
df.head()
```

```python
print(df.isnull())   
```

```python
print(df.isnull().sum())   
```

```python
df['Quantity']=pd.to_numeric(df['Quantity'],errors='coerce')
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())
print(df)
print("Mean Quantity:", df['Quantity'].mean())
```

```python
df['Price Per Unit']=pd.to_numeric(df['Price Per Unit'],errors='coerce')
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Price Per Unit'].median())
print(df)
```

```python
print("Before Forward Fill:")
print(df)
```

```python
df['Location'] = df['Location'].ffill()
print("\nAfter Forward Fill:")
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['Location_Encoded'] = encoder.fit_transform(df['Location'])
print(df)
print(encoder.classes_)
```

```python
encoded_df = pd.get_dummies(df, columns=['Location'])
print(encoded_df)
```

```python
encoded_df = pd.get_dummies(
    df,
    columns=['Location'],
    dtype=int
)
print(encoded_df)
```
